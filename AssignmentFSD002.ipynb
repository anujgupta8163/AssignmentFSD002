Question 1 : What are semantic HTML elements? Why is using them important for web development?

Answer 1 : Semantic HTML elements are HTML tags that clearly describe the meaning and purpose of the content they contain, both to humans and machines.

Using Semantic HTML elements are very important for web development due to several reasons as described below :-
- SEO (Search Engine Optimization) :- Search engines use semantic tags to better understand page content and structure, this improves the indexing and ranking of the content.This improves Search Engine Optimization (SEO).

- Code Readability :- All other developers can quickly understand the code structure because of the semantic HTML elements describing their purpose clearly.

- Future-proofing :- Using semantic HTML elements ensures that your site is more compatible with modern browsers and future technologies.

Question 2: You're designing a blog page. Which semantic elements would you use to structure the page, and why?

Answer 2: To design a blog page, a variety of semantic elements should be used to provide a clear and understanding logical struture for both users and machines.The use of these semantic HTML elements improve the readability and boost the Search Engine Optimization.

Here are the semantic elements I would use to structure the page and why :

- <header> - The header tag is used at the top part of the blog page that contains the blog title and logo.

- <nav> - The nav tag contains the navigation links like Home,Categories,Contact etc.

- <main> - The main tag contains the main content of the blog page.

- <section> - The section tag is used to group related content inside the main page.

- <article> - The article tag is used to represent an article or blog post.

- <footer> - The footer tag is used at the bottom of the page that could include copyright,social links etc.

Question 6: Explain the CSS box model and its components.

Answer 6: The CSS box model describes how every HTML element is represented as a rectangular box on a webpage.CSS box model defines how the content,padding,border and margin work together to determine the total size and spacing of an element.

Following are the CSS Box Model components :-
- Content: The actual text or image inside the element.
- Padding: Space between the content and the border.
- Border: Border wraps the content and the padding.
- Margin: Space outside the border,separating the element from other elements.

Question 7: How do the relative, absolute, and fixed positioning properties differ in CSS?

Answer 7: In CSS, the position property controls how elements are placed. With relative positioning, an element is shifted relative to its original position, but its space in the layout is preserved. With absolute positioning, an element is placed relative to its nearest positioned ancestor (or the page if none exists) and is removed from the normal flow. With fixed positioning, an element is positioned relative to the browser window and remains fixed in place even when the page is scrolled.

Question 8: Write a CSS rule to set a background image for a <div> with the class .banner, ensuring the image covers the entire area without repeating.

Answer 8: Code:

.banner {
  background-image: url("image.jpg");
  background-size: cover;
  background-position: center;
  background-repeat: no-repeat;
}
